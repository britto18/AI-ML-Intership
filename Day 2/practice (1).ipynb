```python
import pandas as pd
data=pd.read_csv('/content/preprocessing_practice_dataset.csv')
```

```python
data.info()
```

```python
data['Age_mean_imputed']=data['Age'].fillna(data['Age'].mean())
data['Monthly_Salary_mean_imputed']=data['Monthly_Salary'].fillna(data['Monthly_Salary'].mean())
```

```python
data['Depatrtment_mode_imputed']=data['Department'].fillna(data['Department'].mode()[0])
data['Education_mode_imputed']=data['Education'].fillna(data['Education'].mode()[0])
data['City_mode_imputed']=data['City'].fillna(data['City'].mode()[0])
```

```python
data
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.set(style="whitegrid")
plt.figure(figsize=(10,6))
sns.barplot(data=data,x="Age",y="Gender",palette='pastel')
plt.title("Age vs Gender")
plt.xlabel("Age")
plt.ylabel("Gender")
```

```python
sns.set(style="whitegrid")
plt.figure(figsize=(10,6))

sns.boxplot(data=data,x="Department",y="Experience_Years",palette='pastel')
plt.title("Department vs Experience")
plt.xlabel("Department")
plt.ylabel("Experience")
```

```python
sns.set(style="whitegrid")
plt.figure(figsize=(10,6))
sns.lineplot(data=data,x="City",y="Gender",palette='pastel')
plt.title("City vs Gender")
plt.xlabel("City")
plt.ylabel("Gender")
```

```python
sns.set(style='whitegrid')
plt.figure(figsize=(10,6))
sns.countplot(data=data,x="Department",palette='pastel')
plt.title("Department Distribution")
plt.xlabel("Department")
plt.ylabel("Count")
```

```python
sns.set(style='whitegrid')
plt.figure(figsize=(10,6))
sns.histplot(data=data,x="Monthly_Salary",palette='pastel')
plt.title("Monthly Salary Distribution")
plt.xlabel("Monthly Salary")
plt.ylabel("Frequency")
```

```python
from sklearn.preprocessing import LabelEncoder
encoded_data=data.copy()
one_hot_encoded=pd.get_dummies(encoded_data['Department'],prefix="Department")
encoded_data=pd.concat([encoded_data,one_hot_encoded],axis=1)
encoded_data
```

```python
performance_class={1:"one",2:"two",3:"three",4:"four",5:"five"}
encoded_data['Performance_Class']=encoded_data['Performance_Rating'].map(performance_class)
encoded_data
```

```python
from sklearn.preprocessing import MinMaxScaler,StandardScaler
min_max_scaler=MinMaxScaler()
standard_scaler=StandardScaler()
encoded_data['Min_Max_Age']=min_max_scaler.fit_transform(encoded_data[['Age']])
encoded_data['standard_Salary']=standard_scaler.fit_transform(encoded_data[['Monthly_Salary']])
encoded_data
```
