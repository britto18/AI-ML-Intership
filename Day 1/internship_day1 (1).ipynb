```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
```

```python
encoded_data=df.copy()
label_encoder=LabelEncoder()
encoded_data['Sex_Label']=label_encoder.fit_transform(encoded_data['Sex'])
```

```python
one_hot_encodede = pd.get_dummies(encoded_data['Embarked'],prefix='Embarked')
encoded_data = pd.concat([encoded_data,one_hot_encodede],axis=1)
```

```python
pclass_mapping = {1:'First',2:'Second',3:'Third'}
encoded_data['Pclass_Category'] = encoded_data['Pclass'].map(pclass_mapping)
```

```python
encoded_data['Pclass_Category'] = encoded_data['Pclass'].map(pclass_mapping)
```

```python
from sklearn.preprocessing import MinMaxScaler,StandardScaler
file_path='/content/titanic uncleaned.csv'
titanic_data=pd.read_csv(file_path)
```

```python
numerical_columns=['Age','Fare','Pclass']
```

```python
titanic_data.dropna(subset=numerical_columns,inplace=True)
```

```python
minmax_scale =  MinMaxScaler()
standard_scale=StandardScaler()
```

```python
titanic_data['Age_MinMax']=minmax_scale.fit_transform(titanic_data[['Age']])
titanic_data['Fare_MinMax']=minmax_scale.fit_transform(titanic_data[['Fare']])
```

```python
titanic_data['Age_Standard']=standard_scale.fit_transform(titanic_data[['Age']])
titanic_data['Fare_Standard']=standard_scale.fit_transform(titanic_data[['Fare']])
```
